```python
# Import the libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 120)

print("Libraries imported successfully!")
```

```python
# Load the dataset
df = pd.read_csv("Nassau Candy Distributor.csv")

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())
```

```python
# Understand the columns
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nFirst 10 records:")
display(df.head(10))

print("\nRandom 5 records:")
display(df.sample(5, random_state=42))
```

```python
# Check missing values
missing_values = df.isnull().sum()

missing_report = pd.DataFrame({
    "Missing Count": missing_values,
    "Missing Percentage": (
        missing_values / len(df) * 100
    ).round(2)
})

display(missing_report)
```

```python
# Check duplicate records
print("Exact duplicate rows:", df.duplicated().sum())

print("Duplicate Row IDs:", df["Row ID"].duplicated().sum())

print("Unique customers:", df["Customer ID"].nunique())

print("Unique orders:", df["Order ID"].nunique())
```

```python
# Check and convert data types
df_clean = df.copy()

df_clean["Order Date"] = pd.to_datetime(
    df_clean["Order Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

df_clean["Ship Date"] = pd.to_datetime(
    df_clean["Ship Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

print(df_clean[["Order Date", "Ship Date"]].dtypes)

display(df_clean[["Order Date", "Ship Date"]].head())
```

```python
# Validate the dates
df_clean["Shipping Delay Days"] = (
    df_clean["Ship Date"] - df_clean["Order Date"]
).dt.days

print("Shipping delay summary:")
display(df_clean["Shipping Delay Days"].describe())

print("\nSample records with dates:")
display(
    df_clean[
        ["Order ID", "Order Date", "Ship Date",
         "Shipping Delay Days"]
    ].head(10)
)
```

```python
# Validate numerical columns
numeric_cols = [
    "Sales", "Units", "Gross Profit", "Cost"
]

print("Negative values in numerical columns:")

for col in numeric_cols:
    print(col, ":", (df_clean[col] < 0).sum())

print("\nSales and cost summary:")
display(df_clean[numeric_cols].describe())
```

```python
# Save the cleaned dataset
output_file = "Nassau_Candy_Cleaned.csv"

df_clean.to_csv(output_file, index=False)

print("Dataset saved as:", output_file)
print("Final shape:", df_clean.shape)
```
